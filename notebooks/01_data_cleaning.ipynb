<h1>E-Commerce Sales Analysis & Demand Prediction</h1>

<h2>1. Data Importing and Cleaning</h2>
<b>1.0. IMPORT REQUIRED LIBRARIES</b>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

<b>1.1. LOAD THE DATASET</b>

```python
file_path = "E:/Project/ecommerce_sales_demand_prediction/data/raw/ecommerce_sales_data.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

<b>1.2. INITIAL DATA INSPECTION</b>

```python
# Display the first 5 rows
print("First 5 Rows:")
display(df.head())

# Display dataset shape
print(f"\nDataset Shape: {df.shape}")

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display data types
print("\nData Types:")
display(df.dtypes)

# Display dataset information
print("\nDataset Information:")
df.info()

# Display basic statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

<b>1.3. CHECK FOR MISSING VALUES</b>

```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
display(missing_values)

print(f"\nTotal Missing Values: {missing_values.sum()}")
```

<b>1.4. CHECK FOR DUPLICATE RECORDS</b>

```python
duplicate_count = df.duplicated().sum()

print(f"Number of Duplicate Records: {duplicate_count}")
```

<b>1.5. CONVERT ORDER DATE TO DATETIME</b>

```python
df["Order Date"]=pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)
print("Order Date converted to datetime successfully.")

print("\nUpdated Data Type:")
print(df["Order Date"].dtype)

print(f"\nInvalid or Missing Dates: {df['Order Date'].isnull().sum()}")
```

<b>1.6. CHECK FOR INVALID VALUES</b>

```python
print("Invalid Value Checks\n")

print(f"Quantity less than or equal to 0: {(df['Quantity'] <= 0).sum()}")
print(f"Sales less than or equal to 0: {(df['Sales'] <= 0).sum()}")
print(f"Profit less than 0: {(df['Profit'] < 0).sum()}")
```

<b>1.7. INSPECT CATEGORICAL VALUES</b>

```python
categorical_columns=[
    "Product Name",
    "Category",
    "Region"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(f"Number of unique values: {df[column].nunique()}")
    print(f"Unique values: {df[column].unique()}")
```

<b>1.8. CREATE USEFUL TIME-BASED FEATURES</b>

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
df["Day of Week"] = df["Order Date"].dt.day_name()

print("Time-based features created successfully.")

display(df.head())
display(df.shape)
```

<b>1.9. CLEAN THE DATASET</b>

```python
# Store the number of rows before cleaning
rows_before_cleaning = df.shape[0]

# Remove duplicate records
df = df.drop_duplicates()

# Remove rows with invalid dates, if any
df = df.dropna(subset=["Order Date"])

# Store the number of rows after cleaning
rows_after_cleaning = df.shape[0]

print("Dataset cleaned successfully.")

print(f"\nRows before cleaning: {rows_before_cleaning}")
print(f"Rows after cleaning: {rows_after_cleaning}")
print(f"Rows removed: {rows_before_cleaning - rows_after_cleaning}")
```

<b>1.10. FINAL DATASET VALIDATION</b>

```python
print("Final Dataset Shape:")
print(df.shape)

print("\nMissing Values:")
display(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
display(df.dtypes)

print("\nFirst 5 Rows of Cleaned Dataset:")
display(df.head())
```

<b>1.11. SAVE THE CLEANED DATASET</b>

```python
output_path = "E:/Project/ecommerce_sales_demand_prediction/data/processed/ecommerce_sales_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(f"File saved to: {output_path}")
```
